```python
# Write a praogram to create a mirror image
input_string = input("Enter srting : ")
n = int(input("enter n :"))

# creating a dictionary for mirror operation

alphabets = "abcdefghijklmnopqrstuvwxyz"
reverse_alphabets = alphabets[ : : -1]
dict1 = dict(zip(alphabets,reverse_alphabets))


# finding the part of the srting on which we do mirror operation

prefix = input_string[0:n-1]
suffix = input_string[n-1:]


# finding the mirror image

mirror = ""
for i in range (0,len(suffix)):
    mirror = mirror + dict1[suffix[i]]

# creating the final string
res = prefix + mirror
print("The result is :",res)
```
