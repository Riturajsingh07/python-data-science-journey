```python
# how to creating a set in python
names={"aman","rohan","mohan"}
print(names)
```

```python
#length of set
print(len(names))
```

```python
# check data type
print(type(names))
```

```python
 #accessing the item of set
for i in names:
  print(i)
```

```python
# check element are present in the se
if "aman" in names:
  print("aman is present in the set")
if "Ram" not in names:
  print("Ram is not present in the set")
```

```python
# add new element in the set
names.add("ram")
print(names)
```

```python
# i we add same element in the set they can't add the element
names.add("aman")
print(names)
```

```python
# add another sequece in a set
names_list = ["sourabh","raj"]
names.update(names_list)
print(names)
print(names)
```

```python
#removing elemnt for the set
names.remove("ram")
print(names)
```

```python

# discard fuction will not throw an errror if value is not present in the set
names.discard("rituraj")
print(names)
```

```python
# joining two set
s1 = {'a','b','c',}
s2 = {'d','e','a'}
print(s1,s2)
# creating a new set
s3 = s1.union(s2)
print(s3)


# another way to joining tow set
s1.update(s2)
print(s1)
```

```python
# keep only duplicate while joining
s1 = {'a','b','c',}
s2 = {'d','e','a'}
s1.intersection_update(s2)  #--> same value are joining to the set
print(s1)
```

```python
# keep all values except duplicate
s1 = {'a','b','c',}
s2 = {'d','e','a'}
s1.symmetric_difference_update(s2)
print(s1)
```
