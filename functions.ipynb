```python
#arbitary argument
def addAllNumber(*args):
  sum = 0
  for i in args:
    sum += i
  return sum
# print("the sum is ",addAllNumber(1,2,3,4,5,))
output = addAllNumber(1,2,3,4,5,)
print("sum is :",output)
```

```python
# Writing a function for calculating sum from 1 to n
def sumOneToN(n):
  sum = 0
  for i in range(1,n+1):
    sum += i
  return sum
n = int(input("Enter n : "))
#function calling
output = sumOneToN(n)
print("sum of all number till n is ",output)
```

```python
# write a python function to calculate the factorial
def factorial(n):
  fact = 1
  if n == 0:
    fact = 1
  else:
    for i in range (1,n+1) :
      fact *= i
  return fact


n = int(input("Enter the number "))
output = factorial(n)
print("The factorial is :",output)
```

```python
def vowel_count(s):
    count = 0
    for ch in s :
      if ch in "aeiou":
          count +=1
    return count
output =  vowel_count("python")
print(output)
```

```python
def count_even(lst):
  count = 0
  for i in lst:
    if i%2==0:
      count+=1
  return count
output = count_even([2,4,6,8,10,12])
print(output)
```
